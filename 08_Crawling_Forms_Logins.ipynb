# Chapter 13: Crawling Through Forms and Logins

```python
import requests
from requests.auth import AuthBase, HTTPBasicAuth
```

## Submitting a Basic Form

```python
params = {
    "firstname": "John",
    "lastname": "Will"
}
resp = requests.post(
    "https://pythonscraping.com/pages/files/processing.php",
    data=params,
)
resp
```

```python
resp.text
```

```python
params = {
    "firstname": "Elliot",
    "lastname": "Robot",
    "email": "mr.robot@mail.me",
    "action": "eclg_add_newsletter",
}
resp = requests.post(
    "https://pythonscraping.com/wp-admin/admin-ajax.php",
    data=params,
)
resp
```

```python
print(resp.text)
```

## Submitting Files and Images

```python
files = {"uploadFile": open("logo.jpg", "rb")}
resp = requests.post(
    "https://pythonscraping.com/pages/files/processing2.php",
    files=files,
)
resp
```

```python
print(resp.text)
```

## Handling Logins and Cookies

```python
params = {"username": "mr.robot", "password": "password"}
resp = requests.post(
    "https://pythonscraping.com/pages/cookies/welcome.php",
    data=params,
)
resp
```

```python
print(resp.text)
```

```python
print(f"Cookie is set to:\n{resp.cookies.get_dict()!r}")
```

```python
print("Going to profile page WITHOUT COOKIES...")
r = requests.get(
    "https://pythonscraping.com/pages/cookies/profile.php",
)
print(r)
print(r.text)
```

```python
print("Going to profile page WITH COOKIES...")
r = requests.get(
    "https://pythonscraping.com/pages/cookies/profile.php",
    cookies=resp.cookies,
)
print(r)
print(r.text)
```

```python
session = requests.Session()

params = {"username": "Mr.Robot", "password": "password"}
s = session.post(
    "https://pythonscraping.com/pages/cookies/welcome.php",
    data=params,
)
print(s)
print(f"Cookies is set to: {s.cookies.get_dict()!r}")
print("Going to profile page...")
s = session.get("https://pythonscraping.com/pages/cookies/profile.php")
print(s)
print(s.text)
```

## HTTP Basic Access Authentication

```python
auth = HTTPBasicAuth("mr.robot", "password")
resp = requests.post(
    "https://pythonscraping.com/pages/auth/login.php",
    auth=auth,
)
print(resp)
print(resp.text)
```
