## PySocks

```python
print(urlopen("https://icanhazip.com").read())
```

```python
resp = requests.get("https://icanhazip.com")
print(resp)
print(resp.text)
```

```python
socks.set_default_proxy(socks.PROXY_TYPE_SOCKS5, "localhost", 9050, rdns=True)
socket.socket = socks.socksocket
```

```python
print(urlopen("https://icanhazip.com").read())
```

```python
resp = requests.get("https://icanhazip.com")
print(resp)
print(resp.text)
```

```python
CHROMEDRIVER_PATH = ChromeDriverManager().install()

chrome_options = Options()
chrome_options.add_argument("--headless")
chrome_options.add_argument("--proxy-server=socks5://127.0.0.1:9050")

driver = webdriver.Chrome(
    service=Service(CHROMEDRIVER_PATH),
    options=chrome_options,
)

driver.get("https://icanhazip.com")
print(driver.page_source)
driver.close()
```

- **DNS leaks:** use `socks.set_default_proxy(socks.PROXY_TYPE_SOCKS5, "localhost", 9050, rdns=True)` so hostnames are resolved by Tor rather than your own machine.
- **Without monkey-patching `socket`:** `pip install requests[socks]` and use `proxies={"http": "socks5h://localhost:9050", "https": "socks5h://localhost:9050"}`. The `h` makes DNS go through Tor, and it only affects that one session instead of every socket in your program.
- **Check it:** hit `https://check.torproject.org/api/ip` and it will tell you whether you're exiting through Tor.
- **Need a new IP?** Restarting the container gives a new circuit, and the control port (`ControlPort` + `NEWNYM`) lets you rotate circuits without restarting.

---

**When encountering any web scraping project, you should always ask yourself:**
* What is the question I want answered or the problem I want solved?
* What data will help me achieve this and where is it?
* How is the website displaying this data? Can I identify exactly which part of the website’s code contains this information?
* How can I isolate the data and retrieve it?
* What processing or analysis needs to be done to make it more useful?
* How can I make this process better, faster, and more robust?
